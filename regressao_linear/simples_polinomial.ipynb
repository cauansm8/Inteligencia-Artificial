Mínimos Quadrados Ordinários (MQO) para Regressão Polinomial (feito em Cálculo Numérico)

$$yi_{prev} = B0 + \sum_{j=1}^{p} Bj Xij$$

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
xi = np.array([-1., 0, 1, 2])
yi = np.array([0., -1, 0, 7])
```

```python
# grau
m = 2
```

```python
# número de pontos
n = len(xi)

# número de coeficientes do polinômio
mn = m + 1

u = np.zeros((n, mn))

# primeira coluna 1, segunda coluna x, terceira coluna x**2
for i in range (0, mn):
    u[:, i] = xi ** i

G = np.zeros((mn, mn))
b = np.zeros(mn)

for i in range (0, mn):
    b[i] = np.sum(u[:, i] * yi)

    for j in range (0, mn):
        G[i,j] = np.sum(u[:, i] * u[:, j])


# resolver o sistema linear
a = np.linalg.inv(G) @ b
print(G)
print(b)
print(a)
```

```python
# erro de truncamento Q
nos = u @ a
Q = np.sum((yi[:] - nos) ** 2)
print(Q)
```
