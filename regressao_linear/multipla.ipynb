Regressão Linear Múltipla

```python
import numpy as np
```

$$ Xb = y$$

$$ Ga = b $$

```python
x = np.array([
    [1, 1],
    [2, 1],
    [1, 2],
    [2, 2]
])

y = np.array([7, 7, 10, 12])

print(x)
```

```python
print(y)
```

```python
quantidade_de_coefs = x.shape[1] + 1

u = np.ones((x.shape[0], quantidade_de_coefs))
u[:, 1:] = x


print(u)
```

$$ G = \begin{bmatrix} <u0, u0> & <u0, u1> & <u0, u2> \\ <u1, u0> & <u1, u1> & <u1, u2> \\ <u2, u0> & <u2, u1> & <u2, u2> \end{bmatrix}$$


```python
g = np.zeros((quantidade_de_coefs, quantidade_de_coefs))

for i in range (g.shape[0]):
    for j in range (g.shape[1]):
        g[i,j] = u[:, i] @ u[:, j]

print(g)
```

$$ b = \begin{bmatrix} <y, u0> \\ <y, u1> \\ <y, u2> \end{bmatrix} $$

```python
b = np.zeros(quantidade_de_coefs)

for i in range (b.shape[0]):
    b[i] = np.sum(u[:, i] * y)

print(b)
```

$$ Ga = b $$

$$ a = G^{-1} b $$

```python
a = np.linalg.inv(g) @ b

print(a)
```

$$ y_{prev} = 1 + 2x_1 + 3x_2 $$

```python
def funcao(x1, x2):
    return 1 + 2*x1 + 3*x2
```

$$ MAE = \frac{1}{n} \sum_{i=1}^{n} | y_{real} - y_{prev} | $$

```python
MAE = 0

for i in range (len(y)):
    MAE += abs(y[i] - funcao(x[i][0], x[i][1]))

MAE = MAE / len(y)

print(f"Mean Absolute Error: {MAE}")
```

$$ MSE = \frac{1}{n} \sum_{i=1}^{n} (y_{real} - y_{prev})^2 $$

```python
MSE = 0

for i in range (len(y)):
    MSE += (y[i] - funcao(x[i][0], x[i][1]))**2

MSE = MSE / len(y)

print(f"Mean Squared Error: {MSE}")
```

$$ RMSE = \sqrt{MSE} $$

```python
print(f"Root Mean Squared Error: {np.sqrt(MSE)}")
```
