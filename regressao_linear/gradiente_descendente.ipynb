Regressão Linear simples com gradiente descendente

$$ \beta^{t+1} = \beta ^ {t} - \alpha \cdot \text{gradiente} J (\beta ^ {t}) $$

$$ J_{abs} = \frac{1}{n} \sum_{i=1}^{n} | e_i | = \text{MAE} $$

$$ J_{quad} = \frac{1}{n} \sum_{i=1}^{n} e_i^2 = \text{MSE} $$

$$ \alpha = [0, 1] $$

$$ \text{gradiente} = \text{derivada} $$

```python
import numpy as np
```

```python
alpha = 0.001
```

$$ J_{quad} = \frac{1}{n} \sum_{i=1}^{n} e_i^2 $$

$$ \text{derivada} = \text{MSE}^2 = \frac{2}{n} \cdot \text{Resíduo} $$


```python
x = np.array([1, 3, 5, 7, 10, 12, 15, 20])
y = np.array([3.5, 5.2, 7.8, 9.1, 12.5, 13.0, 18.2, 21.0])
```

```python
epocas = 100
```

```python
b0 = b1 = 0

for i in range (epocas):
    y_prev = b0 + b1 * x

    residuo = y - y_prev

    derivada_b0 = -2 / len(y) * np.sum(residuo)

    derivada_b1 = -2 / len(y) * np.sum(residuo * x)

    b0 = b0 - alpha * derivada_b0
    b1 = b1 - alpha * derivada_b1

b0 = round(b0, 2)
b1 = round(b1, 2)
```

```python
print(f"y_prev = {b0} + {b1} x")
```

$$ MAE = \frac{1}{n} \sum_{i=1}^{n} | y_{real} - y_{prev} | $$

```python
MAE = 0

for i in range (len(y)):
    MAE += abs(y[i] - y_prev[i])

MAE = MAE / len(y)

print(f"Mean Absolute Error: {MAE}")
```

$$ MSE = \frac{1}{n} \sum_{i=1}^{n} (y_{real} - y_{prev})^2 $$

```python
MSE = 0

for i in range (len(y)):
    MSE += (y[i] - y_prev[i]) ** 2

MSE = MSE / len(y)

print(f"Mean Squared Error: {MSE}")
```

$$ RMSE = \sqrt{MSE} $$

```python
RMSE = np.sqrt(MSE)

print(f"RMSE: {RMSE}")
```
