Mínimos Quadrados Ordinários (MQO):

$$ B1 = \frac{\sum_{i=1}^{n} (xi - x_{media}) (yi - y_{media})}{\sum_{i=1}^{n} (xi - x_{media})^2}$$

$$ B0 = y_{media} - B1 x_{media}$$

$$yi_{prev} = B0 + B1 xi $$

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.array([1, 2, 3, 4])
y = np.array([2, 3, 5, 4])
```

```python
x_media = np.mean(x)
y_media = np.mean(y)
```

```python
em_cima = 0
em_baixo = 0

for i in range (len(x)):
    em_cima += (x[i] - x_media) * (y[i] - y_media)
    em_baixo += (x[i] - x_media)**2

B1 = em_cima / em_baixo
```

```python
B0 = y_media - B1 * x_media
```

```python
print(f"y_pred = {B0} + {B1}x")

y_prev = B0 + x * B1

print(y_prev)
```

```python
plt.scatter(
    x, y, color="blue", s=80, zorder=5, label="Dados Reais (Y)"
)

plt.plot(
    x,
    y_prev,
    color="red",
    linestyle="--",
    linewidth=2,
    label=f"Reta Ajustada: {{Y_previsão}} = {B0:.1f} + {B1:.1f}X",
)

for i in range(len(x)):
    plt.vlines(
        x[i],
        ymin=min(y[i], y_prev[i]),
        ymax=max(y[i], y_prev[i]),
        colors="gray",
        linestyles="dotted",
    )

plt.title("Regressão Linear Simples: Dados Observados vs. Ajustados", fontsize=12)
plt.xlabel("Variável Explicativa (X)", fontsize=10)
plt.ylabel("Variável Resposta (Y)", fontsize=10)
plt.xticks(x)
plt.grid(True, linestyle=":", alpha=0.6)
plt.legend(fontsize=10)

plt.show()
```

$$ MAE = \frac{1}{n} \sum_{i=1}^{n} | y_{real} - y_{prev} | $$

```python
soma = 0
for i in range (len(y)):
    soma += abs(y[i] - y_prev[i])

MAE = soma / len(x)

print(f"MAE: {MAE}")
```

$$ MSE = \frac{1}{n} \sum_{i=1}^{n} (y_{real} - y_{prev})^2 $$

```python
soma = 0
for i in range (len(y)):
    soma += (y[i] - y_prev[i])**2

MSE = soma / len(x)

print(f"MSE: {MSE}")
```

$$ RMSE = \sqrt{MSE} $$

```python
RMSE = np.sqrt(MSE)

print(f"RMSE: {RMSE}")
```

Soma dos quadrados total (SST)

$$ SST = \sum_{i=1}^{n} (yi_{real} - yi_{media})^2 $$

```python
SST = 0

for i in range (len(y)):
    SST += (y[i] - y_media)**2

print(f"SST: {SST}")
```

Soma dos quadrados dos erros (SSE)

$$ SSE = \sum_{i=1}^{n} (yi_{real} - yi_{prev})^2 $$

```python
SSE = 0

for i in range (len(y)):
    SSE += (y[i] - y_prev[i])**2

print(f"SSE: {SSE}")
```

Soma dos quadrados da regressão (SSR)

$$ SSR = \sum_{i=1}^{n} (yi_{prev} - y_{media})^2 $$

```python
SSR = 0

for i in range (len(y)):
    SSR += (y_prev[i] - y_media)**2

print(f"SSR: {SSR}")
```

Coeficiente de Determinação ($R^2$)

$$ R^2 = 1 - \frac{SSE}{SST} $$

```python
R_quadrado = 1 - SSE/SST

print(f"R^2: {R_quadrado}")
```
